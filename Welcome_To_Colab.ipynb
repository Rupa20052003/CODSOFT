```
def tenure_lab(Bank):
    if Bank["Tenure"] <= 6:
        return "Tenure_0_6"

    elif (Bank["Tenure"] > 6) & (Bank["Tenure"] <= 12):
        return "Tenure_7_12"

    elif (Bank["Tenure"] > 12) & (Bank["Tenure"] <= 24):
        return "Tenure_13_24"

    elif (Bank["Tenure"] > 24) & (Bank["Tenure"] <= 36):
        return "Tenure_25_36"

    elif Bank["Tenure"] > 36:
        return "Tenure_gt_36"

# Bank = x
Bank["Tenure_group"] = Bank.apply(lambda x: tenure_lab(x), axis=1)  #X = bank
```
